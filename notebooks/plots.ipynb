```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_file = "DETR-Report-Train-Lr0.0001-WeightDecay0.0001-Params41589209.csv"
val_file = "DETR-Report-Val-Lr0.0001-WeightDecay0.0001-Params41589209.csv"

df_train = pd.read_csv(train_file)
df_val = pd.read_csv(val_file)
```

```python
df_train.head(-1)
```

```python
df_val.head(-1)
```

```python
n_cols, n_rows = 3, 1
figsize = (3.5 * n_cols, 4.5 * n_rows)
fig, axes = plt.subplots(ncols=n_cols, figsize=figsize)
axes = axes.flatten()

columns = {
    "loss": "DETR loss", 
    "loss_ce": "Cross-Entropy loss", 
    "loss_bbox": "Bounding-Box loss"
}
for col, ax in zip(columns.keys(), axes):
    ax.plot(df_train["epoch"], df_train[col], label="Train")
    ax.plot(df_val["epoch"], df_val[col], label="Val")
    ax.set_xlabel("Epochs")
    ax.set_ylabel(columns[col])
    ax.set_box_aspect(1)  
    ax.legend()

plt.tight_layout()
```
